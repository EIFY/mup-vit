```python
import json
import os

import numpy as np

ROOT = "/media/jason-chou/T31/imagenet-runs/logs"

# Slightly simplified https://github.com/huggingface/pytorch-image-models/blob/main/timm/data/real_labels.py

class RealLabelsImagenet:

    def __init__(self, filenames, real_labels, topk=(1, 5)):
        real_labels = {f'ILSVRC2012_val_{i + 1:08d}.JPEG': labels for i, labels in enumerate(real_labels)}
        self.real_labels = real_labels
        self.filenames = filenames
        assert len(self.filenames) == len(self.real_labels)
        self.topk = topk
        self.is_correct = {k: [] for k in topk}
        self.sample_idx = 0

    def add_result(self, output):
        maxk = max(self.topk)
        _, pred_batch = output.topk(maxk, 1, True, True)
        pred_batch = pred_batch.cpu().numpy()
        for pred in pred_batch:
            filename = self.filenames[self.sample_idx]
            filename = os.path.basename(filename)
            if self.real_labels[filename]:
                for k in self.topk:
                    self.is_correct[k].append(
                        any([p in self.real_labels[filename] for p in pred[:k]]))
            self.sample_idx += 1

    def get_accuracy(self, k=None):
        if k is None:
            return {k: float(np.mean(self.is_correct[k])) for k in self.topk}
        else:
            return float(np.mean(self.is_correct[k]))
```

```python
import torch
from simple_vit import SimpleVisionTransformer
import torchvision.datasets as datasets
from torchvision.transforms import v2
from scipy.stats import entropy

def eval_best(p, ds, real_labels, device):
    entropy_file = 'val_entropy.npy'
    logit_file = 'val_logits.npy'
    acc_file = 'acc.json'
    if os.path.exists(acc_path := os.path.join(p, acc_file)):
        print(p + ' already done!')
        with open(acc_path) as f:
            acc = json.load(f)
    else:
        print(p)
        hidden_dim = 384
        input_resolution = 224
        
        vit = SimpleVisionTransformer(
            image_size=input_resolution,
            patch_size=16,
            num_layers=12,
            num_heads=6,
            hidden_dim=hidden_dim,
            mlp_dim=hidden_dim * 4,
            representation_size=hidden_dim
        )
        acc = {}
        best_path = os.path.join(p, 'checkpoints/model_best.pth.tar')
        ckpt = torch.load(best_path, weights_only=True)
        acc['best_acc1'] = ckpt['best_acc1']
        try:
            vit.load_state_dict(ckpt['state_dict'])
        except:
            vit.load_state_dict({k[len('module.'):]: v for k, v in ckpt['state_dict'].items()})
    
        vit = vit.to(device)
        val_logits = []
        result = []
    
        for _, batch in enumerate(ds):
            images, target = batch
            images = images.to(device)
            target = target.to(device)
            images.requires_grad = True
            logits, _ = vit(images, 1.0, target, target)
            real_labels.add_result(logits)
            val_logits.append(logits.detach().cpu().numpy())
            target_logits = logits[torch.arange(logits.shape[0]), target]
            total = sum(target_logits)
            total.backward()
            grad_sum_clipped = torch.clamp(images.grad.sum(dim=1), min=0.)
            grad_sum_clipped_normalized = grad_sum_clipped / grad_sum_clipped.sum(dim=(1, 2), keepdim=True)
            grad_sum_clipped_normalized = grad_sum_clipped_normalized.flatten(start_dim=1).cpu().numpy()
            result.append(entropy(grad_sum_clipped_normalized, axis=1))
    
        acc['best_real_acc1'], acc['best_real_acc5'] = real_labels.get_accuracy(k=1), real_labels.get_accuracy(k=5)
        val_logits = np.concatenate(val_logits, dtype=np.float32)
        np.save(os.path.join(p, logit_file.split('.')[0]), val_logits)    
        val_entropy = np.concatenate(result, dtype=np.float32)
        np.save(os.path.join(p, entropy_file.split('.')[0]), val_entropy)
        with open(acc_path, 'w') as f:
            json.dump(acc, f)

    print(acc)
    return
```

```python
device = torch.device("cuda")

def torch_ds():
    input_resolution = 224
    data = "/data/ImageNet/"
    value_range = v2.Normalize(
        mean=[0.5] * 3,
        std=[0.5] * 3
    )
    
    val_dataset = datasets.ImageNet(
        data,
        split='val',
        transform=v2.Compose([
            v2.ToImage(),
            v2.Resize(256),
            v2.CenterCrop(input_resolution),
            v2.ToDtype(torch.float32, scale=True),
            value_range,
        ])
    )

    batch_size = 64

    val_loader = torch.utils.data.DataLoader(
        val_dataset, batch_size=batch_size, shuffle=False,
        num_workers=4, pin_memory=True, sampler=None,
        multiprocessing_context='spawn'
    )
    return val_loader
```

```python
run = "300ep-Scion"
ds = torch_ds()
real_labels = RealLabelsImagenet(filenames, raw_real_labels)
eval_best(os.path.join(ROOT, run), ds, real_labels, device)
```

```python
import statistics, itertools
from scipy.stats import ttest_ind

def epoch_entropy_scatter(filename, field='best_acc1'):
    epoch = [30, 60, 90, 150, 300]
    data = {ep: [] for ep in epoch}
    
    for run in os.listdir(ROOT):
        if os.path.isdir(os.path.join(ROOT, run)) and (run in scion_runs or ('Scion' in run and 'lower-scale' in run)):
            for ep in epoch:
                if str(ep) + 'ep' in run:
                    arr = np.load(os.path.join(ROOT, run, filename))
                    if np.isnan(arr).any():
                        print(run, np.sum(np.isnan(arr)))
                        arr = arr[~np.isnan(arr)]
                    path = os.path.join(ROOT, run, "acc.json")
                    with open(path) as f:
                        d = json.load(f)
                    data[ep].append((d[field], np.mean(arr)))
                    break
    for ep, seq in data.items():
        x, y = zip(*seq)
        plt.scatter(x, y, label=f"{ep}ep")
    
    prop_cycle = plt.rcParams['axes.prop_cycle']
    colors = prop_cycle.by_key()['color']
    ys = []
    
    for (ep, seq), color in zip(data.items(), colors):
        x, y = zip(*seq)
        ys.append(y)
        y_bar = statistics.mean(y)
        plt.hlines(y=y_bar, xmin=min(x), xmax=max(x),
                   color=color, linestyle='dashed')
        print(ep, y_bar)
    
    print(len(ys))
    for pair in itertools.combinations(ys, r=2):
        a, b = pair
        print(ttest_ind(a, b, equal_var=False))

epoch_entropy_scatter('val_entropy.npy')
plt.legend(loc="lower left")
plt.xlabel("Top-1 val. accuracy")
plt.ylabel("Mean Gradient Entropy")
plt.savefig("scion_grad_entropy_vs_acc.png")
plt.show()
```

```python
import itertools, os, json
from scipy.stats import ttest_ind

import numpy as np
import matplotlib
from cycler import cycler

import matplotlib.pyplot as plt

all_wd = list(range(1, 5))
old_cycle = matplotlib.rcParams['axes.prop_cycle']
matplotlib.rcParams['axes.prop_cycle'] = cycler('color', plt.cm.Spectral(np.linspace(0, 1, len(all_wd))))
data = {f"{wd}x-wd": [] for wd in all_wd}
count = 0

for run in os.listdir(ROOT):
    if os.path.isdir(os.path.join(ROOT, run)) and (run in scion_runs or ('Scion' in run and 'lower-scale' in run)):
        count += 1
        for wd in all_wd:
            if (key := f"{wd}x-wd") in run:
                arr = np.load(os.path.join(ROOT, run, 'val_entropy.npy'))
                path = os.path.join(ROOT, run, "acc.json")
                with open(path) as f:
                    d = json.load(f)
                data[key].append((d['best_acc1'], np.mean(arr)))
                break
        else:
            wd = 1
            key = f"{wd}x-wd"
            arr = np.load(os.path.join(ROOT, run, 'val_entropy.npy'))
            path = os.path.join(ROOT, run, "acc.json")
            with open(path) as f:
                d = json.load(f)
            data[key].append((d['best_acc1'], np.mean(arr)))

ls_y = []
for s, seq in data.items():
    x, y = zip(*seq)
    ls_y.append((s, y))
    plt.scatter(x, y, label=str(s))

print(len(ls_y))
for pair in itertools.combinations(ls_y, r=2):
    (ls1, a), (ls2, b) = pair
    print((ls1, ls2), ttest_ind(a, b, equal_var=False))

print(f"{count=}")
plt.legend(loc="lower left")
plt.xlabel("Top-1 val. accuracy")
plt.ylabel("Mean Gradient Entropy")
plt.savefig("scion_grad_entropy_vs_wd.png")
plt.show()
matplotlib.rcParams['axes.prop_cycle'] = old_cycle
```

```python
lines = []
l = ['', 'baseline', 'optimized']
lines.append(' & '.join(l))

def read(template, file, field, n=3):
    l = []
    for i in range(3):
        path = os.path.join(template + f"-{i}", file)
        if not i and not os.path.exists(path):
            path = os.path.join(template, file)
        with open(path) as f:
            d = json.load(f)
        l.append(d[field])
    return statistics.mean(l), statistics.stdev(l)

def format_acc(acc):
    l = [f'{100 * x:.2f}' for x in acc]
    if len(acc) > 1:
        max_i = max(range(len(acc)), key=lambda i: acc[i])
        l[max_i] = "\\textbf{" + l[max_i] + "}"
    return l

lower_scale = [0.25, 0.25, 0.1, None, 0.025]

for scale, ep in zip(lower_scale, epoch):
    acc = []
    template = ROOT + f"/{ep}ep-Scion-2x-wd"
    acc.append(read(template, file="acc.json", field="best_acc1"))
    if scale is not None:
        template = ROOT + f"/{ep}ep-Scion-2x-wd-{scale}-lower-scale"
        acc.append(read(template, file="acc.json", field="best_acc1"))
    mean, stdev = zip(*acc)
    mean_str = format_acc(mean)
    l = [f'{ep}ep'] + [m + f'$\pm {100 * s:.2f}$' for m, s in zip(mean_str, stdev)]
    if scale is None:
        l.append('---')
    lines.append(' & '.join(l))

for line in lines:
    print(line + ' \\\\')
    print('\\hline')
```

```python
field = "best_real_acc1"
lines.clear()

for scale, ep in zip(lower_scale, epoch):
    acc = []
    template = ROOT + f"/{ep}ep-Scion-2x-wd"
    acc.append(read(template, file="acc.json", field=field))
    if scale is not None:
        template = ROOT + f"/{ep}ep-Scion-2x-wd-{scale}-lower-scale"
        acc.append(read(template, file="acc.json", field=field))
    mean, stdev = zip(*acc)
    mean_str = format_acc(mean)
    l = [f'{ep}ep'] + [m + f'$\pm {100 * s:.2f}$' for m, s in zip(mean_str, stdev)]
    if scale is None:
        l.append('---')
    lines.append(' & '.join(l))

for line in lines:
    print(line + ' \\\\')
    print('\\hline')
```
